```python
import pandas as pd
import matplotlib.pyplot as plt
history = pd.read_csv('reports/cnn_training_history.csv')
display(history.groupby('run')['val_macro_f1'].max().sort_values(ascending=False))
for run, group in history.groupby('run'):
    plt.plot(group.epoch, group.val_macro_f1, marker='o', label=run)
plt.xlabel('Epoch'); plt.ylabel('Validation macro F1'); plt.legend(); plt.tight_layout()
```
